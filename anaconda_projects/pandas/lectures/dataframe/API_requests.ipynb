```python
import requests
import pandas as pd

url = "https://stephen-king-api.onrender.com/api/books"
res = requests.get(url)

print("Status:", res.status_code)

data = res.json()

# Convert to DataFrame
df = pd.json_normalize(data)

print(df.head())
```

```python
print(df.columns)
```

```python
df = pd.json_normalize(data["data"])
print(df.columns)
```
