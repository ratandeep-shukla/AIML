```python
import pandas as pd
```

```python
#dataframe as dictionary
info = {
    "Name" : ["Rahul", "Harshita", "Shrishti"],
    "CGPA" : [7.4, 8.3, 9.5]
}
df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
```

```python
#dataframe as list
df = pd.DataFrame([["Adam",23],["Bob",23],["Eve",23]],columns=["Name","Age"])
print(df)
```

```python
#dataframe as numpy_arr
import numpy as np
np_arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_arr,columns=["A","B","C"])
print(df)
```
