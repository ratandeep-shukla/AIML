```python
#reshaping
import numpy as np
arr = np.array([[1,2,3],[4,5,6]])
print(arr,arr.shape)
reshaped = arr.reshape((3,2))
print(reshaped,reshaped.shape)
```

```python
flat = arr.flatten()
print(flat,flat.shape)
```

```python
#fancy indexing
arr = np.array([1,2,3,4,5,6])
idx = [0,2,3]
print(arr[idx])
```

```python
#boolean indexing
arr = np.array([1,2,3,4,5,7])
print(arr[arr>2])
print(arr[arr%2==0])
print(arr[arr%2!=0])
```

```python
#slicing
arr = np.array([1,2,3,4,5,6])
print(arr[1:4])
print(arr[1:])
print(arr[:4])
print(arr[::2])
```

```python
#copy vs view
nums = [1,2,3,4,5]
sub_list = nums[1:3]
sub_list[0] = 100
print(nums)
print(sub_list)

arr = np.array(nums)
sub_arr = arr[1:3]
sub_arr2 = arr[1:3].copy()
sub_arr[0] = 100
sub_arr2[1] = 200
print(arr)
print(sub_arr)
print(sub_arr2)
```
