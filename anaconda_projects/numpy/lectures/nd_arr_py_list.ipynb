```python
import numpy as np
import time
```

```python
#execution performance
size = 10000000
#python list
py_list = list(range(size))
start1 = time.time()
sq_list = [x**2 for x in py_list]
end1 = time.time()
print(f"python list time = {end1-start1} seconds")

#numpy arrays
np_arr = np.array(py_list)
start2 = time.time()
sq_arr = np_arr ** 2
end2 = time.time()
print(f"numpy array time = {end2-start2} seconds")
```

```python
#memory
import sys
print(f"python list size = {sys.getsizeof(py_list)*len(py_list)} bytes")
print(f"numpy arraysize = {np_arr.nbytes} bytes")
```
